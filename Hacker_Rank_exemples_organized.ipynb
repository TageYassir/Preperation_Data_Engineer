# 1. Sherlock and the Valid String

**Topic:** Frequency counting · Edge cases

## Problem overview

A string is considered **valid** when every distinct character appears the same number of times. It is also valid if removing **at most one character occurrence** makes all remaining character frequencies equal.

Complete `isValid(s)` so that it returns:

- `YES` if the string is valid.
- `NO` otherwise.

## Examples

| Input string | Result | Reason |
|---|---|---|
| `aabbcc` | `YES` | Every character appears twice. |
| `aabbccc` | `YES` | Removing one `c` leaves every character with frequency 2. |
| `aabbcd` | `NO` | One deletion cannot make all character frequencies equal. |
| `aabbccddeefghi` | `NO` | Several characters occur once, so a single deletion is not enough. |

## HackerRank sample cases

**Sample 0**

```text
aabbcd
```

Expected output:

```text
NO
```

**Sample 1**

```text
aabbccddeefghi
```

Expected output:

```text
NO
```

**Sample 2**

```text
abcdefghhgfedecba
```

Expected output:

```text
YES
```

## Function contract

- **Function:** `isValid(s)`
- **Parameter:** `s` — the input string
- **Returns:** the string `YES` or `NO`

## What to pay attention to

The key is to compare **character frequencies**, not character positions. Test both the case where all frequencies already match and the cases where one deletion could—or could not—resolve the mismatch.

````python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'isValid' function below.
#
# The function is expected to return a STRING.
# The function accepts STRING s as parameter.
#

def isValid(s):
    seen = {}

    for i,c in enumerate(s):
        # print("char is : ",c)
        seen[c] = seen.get(c,0)+1
       #print(f"seen at {i} ({c}) = {seen[c]}")
        
    frequency_of_values = {}
    for frequency in seen.values():
        frequency_of_values[frequency] = frequency_of_values.get(frequency,0) + 1
    
    print(frequency_of_values)
    
    if len(frequency_of_values) == 1:
        return "YES"
    
    if len(frequency_of_values) == 2:
        (f1, c1),(f2, c2) = frequency_of_values.items()
        
        if (f1 == 1 and c1 == 1) or (f2 == 1 and c2 == 1):
            return "YES"
        
        if abs(f1 - f2) == 1 and min(c1,c2) == 1:
            return "YES"
    
    return "NO"
````

````python
if __name__ == '__main__':
    s = "abcdeff"
    result = isValid(s)
    print(result)
````

# 2. Sherlock and Anagrams

**Topic:** Substrings · Hash maps · Combinatorics

## Problem overview

Two strings are anagrams if one can be rearranged to form the other. Given a string `s`, count the **unordered pairs of substrings** that are anagrams of one another.

Substrings must come from different positions; identical substring text can still form a pair when it occurs at multiple positions.

## Examples

- For `abba`, the answer is `4`.
- For `abcd`, the answer is `0` because no two substrings form an anagram pair.
- For `kkkk`, the answer is `10`: every pair of positions for equal-length substrings contributes to the total.

## Function contract

- **Function:** `sherlockAndAnagrams(s)`
- **Parameter:** `s` — a string containing lowercase English letters
- **Returns:** an integer representing the number of anagrammatic substring pairs

## Useful observation

Substrings with the same character composition share the same canonical signature. Grouping substrings by signature makes it possible to count pairs within each group.

````python
def sherlockAndAnagrams(s):
    n = len(s)
    signatures = {}
    
    
    for i in range(n):
        for j in range(i,n):
            print(f"This is before processing i={i}, j={j}, s[i:j+1]={s[i:j+1]}")
            print(f"this is the sorted version: {sorted(s[i:j+1])}")
            sig = ''.join(sorted(s[i:j+1]))
            print(f"this is the signature: {sig}")
            signatures[sig] = signatures.get(sig, 0) + 1

    total_pairs = 0
    for sig, count in signatures.items():
        if count > 1:
            total_pairs += (count * (count - 1)) // 2
            
    print(f"signatures={signatures}")
    print(f"total_pairs={total_pairs}")
    return total_pairs
````

````python
s="abba"
result = sherlockAndAnagrams(s)
````

# 3. Sherlock and Cost

**Topic:** Dynamic programming · Greedy choices at boundaries

## Problem overview

You are given an integer array `B`. Construct an array `A` such that, for every index `i`,

\[
1 \le A_i \le B_i
\]

Choose the values of `A` to maximize the total cost:

\[
\text{Cost}(A) = \sum_{i=2}^{n} |A_i - A_{i-1}|
\]

In other words, maximize the sum of absolute differences between adjacent elements.

## Small illustration

For `B = [1, 2, 3]`, each position in `A` can take any integer from `1` through the corresponding value in `B`. Some valid choices are:

- `[1, 1, 1]`
- `[1, 2, 1]`
- `[1, 2, 3]`

Each choice produces a different cost. The goal is to find the maximum possible one.

## HackerRank sample

**Input**

```text
1
5
10 1 10 1 10
```

**Expected output**

```text
36
```

The maximum is obtained by alternating between the high and low values: `10, 1, 10, 1, 10`.

## Function contract

- **Function:** `cost(B)`
- **Parameter:** `B` — an array of integers
- **Returns:** the maximum achievable cost

## Dynamic-programming perspective

At each position, consider the best total cost when the current chosen value is at its **low boundary** (`1`) and at its **high boundary** (`B[i]`). Update these two states as you move through the array, rather than trying every possible array `A`.

````python
def cost(B):
    low = 0
    high = 0


    for i in range(1, len(B)):
        prev_B = B[i-1]
        cur_B = B[i]

        new_low = max(low, high + (prev_B -1)) # Can only go down by 1, so the new low is either the previous low or the previous high minus 1
        new_high = max(low + (cur_B -1), high + abs(cur_B - prev_B))

        low = new_low
        high = new_high

    return max(low, high)
````

````python
cost([1, 2, 3, 2, 1])
````

# 4. Minimum Swaps 2

**Topic:** Arrays · Index placement · Swap counting

## Problem overview

You are given an array containing the consecutive integers from `1` to `n`, with no duplicates. Find the **minimum number of swaps** needed to sort the array in ascending order. You may swap any two elements.

| 3 | `[1, 2, 3, 4, 5, 7, 6]` | `(4, 5)` |

This example takes **5 swaps** to sort.

## HackerRank sample cases

## Function contract

- **Function:** `minimumSwaps(arr)`
- **Parameter:** `arr` — a permutation of the integers from `1` to `n`
- **Returns:** the minimum swap count

## Key observation

The value `x` belongs at index `x - 1` in a zero-indexed Python list. Placing each value in its correct position avoids sorting the array repeatedly.

````python
def minimumSwaps(arr):
    swaps = 0
    n = len(arr)
    i = 0

    while i < n:
        if arr[i] != i + 1:
            home = arr[i] - 1        
            arr[i], arr[home] = arr[home], arr[i]
            swaps += 1
        else:
            i += 1

    return swaps
````

````python
arr = [4, 3, 1, 2, 5, 7, 6]
result = minimumSwaps(arr)
print(f"minimumSwaps result: {result}")
````

## Function contract

- **Function:** `freqQuery(queries)`
- **Parameter:** `queries` — a list of `[operation, value]` pairs
- **Returns:** the results for all type-`3` queries, in order

## Design challenge

A straightforward implementation can maintain each value's count, but checking all counts for every type-`3` query can be costly. The next section introduces a faster frequency-based approach.

````python
def freqQuery(queries):
    list_of_values = []
    freq = {}
    for q in queries:
        command, value = q
        if command == 1:
            list_of_values.append(value)
            freq[value] = freq.get(value, 0) + 1
        if command == 2:
            if value in list_of_values:
                list_of_values.remove(value)
                freq[value] = freq.get(value, 0) - 1
        if command == 3:
            if value in freq.values():
                print(1)
            else:
                print(0)
````

````python
freqQuery([[1, 5], [1, 6], [3, 2], [1, 10], [1, 10], [1, 6], [2, 5], [3, 2]])
````

### ⚡ Frequency Queries — Faster Version

The optimized implementation keeps **two dictionaries** so type-`3` queries can be answered without scanning every value:

- `count[value]` → how many times a particular value appears.
- `freq[frequency]` → how many distinct values currently have that frequency.

When a value's count changes, update both dictionaries: remove the value from its old frequency bucket and add it to its new bucket (when the new frequency is positive). A query `3 z` can then be answered by checking whether `freq[z]` is greater than zero.

This avoids repeatedly scanning the full collection for frequency queries.

````python
def freqQuery(queries):
    count = {}   # value -> frequency
    freq = {}    # frequency -> how many values have it
    result = []

    for op, val in queries:
        if op == 1:
            old_f = count.get(val, 0)
            new_f = old_f + 1
            count[val] = new_f
            freq[old_f] = freq.get(old_f, 0) - 1
            freq[new_f] = freq.get(new_f, 0) + 1
        elif op == 2:
            old_f = count.get(val, 0)
            if old_f > 0:
                new_f = old_f - 1
                count[val] = new_f
                freq[old_f] -= 1
                if new_f > 0:
                    freq[new_f] = freq.get(new_f, 0) + 1
        else:
            result.append(1 if freq.get(val, 0) > 0 else 0)

    return result
````

````python
freqQuery([[1, 5], [1, 6], [3, 2], [1, 10], [1, 10], [1, 6], [2, 5], [3, 2]])
````

# 6. Max Array Sum

**Topic:** Dynamic programming · Non-adjacent selection

## Problem overview

Given an array of integers, find the maximum sum of a subset that contains **no adjacent elements**. The empty subset is allowed, so if every element is negative, the maximum sum may be `0`.

You may choose a single element, multiple non-adjacent elements, or no elements at all.

## Example

For the array `[-2, 1, 3, -4, 5]`, one valid non-adjacent subset is `[3, 5]`, with sum `8`.

For the smaller illustration below, the maximum among the listed subsets is shown:

The best sum from this illustration is `8`.

## HackerRank sample cases

## Function contract

- **Function:** `maxSubsetSum(arr)`
- **Parameter:** `arr` — an array of integers
- **Returns:** the maximum sum obtainable from a non-adjacent subset

## Dynamic-programming perspective

For each position, compare two choices: skip the current element, or take it and add it to the best sum ending two positions earlier. Keeping only the previous two states is enough; the complete table does not need to be stored.

````python
def maxSubsetSum(arr):
    n = len(arr)
    
    if n == 0:
        return 0
    elif n == 1:
        return max(0, arr[0])
    
    prev2 = max(0, arr[0])
    prev1 = max(prev2, arr[1]) 

    for i in range(2, n):
        curent = max(prev1, prev2 + arr[i])
        prev2 = prev1
        prev1 = curent

    return prev1
````

# 7. Count Triplets

**Topic:** Hash maps · Counting combinations · Geometric progression

## Problem overview

Given an array `arr` and a common ratio `r`, count triplets of indices `i < j < k` whose values form a geometric progression:

\[
arr[j] = arr[i] \times r, \qquad arr[k] = arr[j] \times r
\]

The order of indices matters, and repeated values at different positions can create different triplets.

## HackerRank sample cases

**Sample 0**

```text

```

Expected output:

```text

```

**Sample 1**

```text

1 3 9 9 27 81
```

Expected output:

```text

```

**Sample 2**

```text
5 5
1 5 5 25 125
```

Expected output:

```text
4
```

## Function contract

- **Function:** `countTriplets(arr, r)`
- **Parameters:** `arr` — the integer array; `r` — the common ratio
- **Returns:** the number of valid triplets

## Core counting idea

Treat each array element in turn as the **middle value** of a possible triplet. Count valid matching values to its left and right, then multiply those counts to find how many triplets use that middle position.

````python
def countTriplets(arr, r):
    right = {}

    for val in arr:
        print(f"val={val}, right={right}")
        right[val] = right.get(val, 0) + 1

    print(f"Initial right={right}")

    left = {}
    total = 0

    for val in arr:
        right[val] = right.get(val, 0) - 1
        print(f"This value was just processed: {val}, updated right={right}")
        if val % r == 0:
            print(f"\nval={val} is divisible by r={r}, checking left and right counts")
            left_count = left.get(val // r, 0)
            print(f"left_count for {val // r} = {left_count}")  
            right_count = right.get(val * r, 0)
            print(f"right_count for {val * r} = {right_count}")
            total += left_count * right_count
            print(f"Updated total={total} after processing val={val}\n")
        left[val] = left.get(val, 0) + 1

    return total
````

````python
countTriplets([1, 2, 2, 4], 2)
````

## Count Triplets — Step-by-Step Walkthrough

This section explains the two-dictionary strategy used to count geometric-progression triplets in one pass.

### 1. The big picture

A valid triplet has three ordered positions, `i < j < k`, with values:

\[
x/r,\quad x,\quad x\times r
\]

Here, `x` is the middle value. For each element in the array, ask: **How many valid left partners and right partners can pair with this value?** Multiply those counts and add the result to the running total.

### 2. The two dictionaries

- **`right` — values still ahead:** initialize it with the frequency of every value in the entire array. As you move from left to right, subtract the current value from this dictionary because it is no longer ahead of the current position.
- **`left` — values already passed:** start with an empty dictionary. After processing the current value, add it to `left` so it can serve as a left partner for later positions.

### 3. Process each value in this order

**Step 1 — Remove the current value from `right`.**

The current position must not count as a right partner for itself.

**Step 2 — Count triplets with the current value as the middle.**

For a current value `x`, the desired left value is `x / r` and the desired right value is `x * r`. When `x` is divisible by `r`, look up:

- `left_count = left[x // r]`
- `right_count = right[x * r]`

The contribution from this middle position is:

\[
\text{contribution} = \text{left\_count} \times \text{right\_count}
\]

Add that contribution to the running total. If either count is zero, this middle position contributes no triplets.

**Step 3 — Add the current value to `left`.**

It is now behind the next position, so increase its count in `left`.

### 4. Why the order matters

The processing order must be:

1. Decrement the current value in `right`.
2. Count triplets using the current value as the middle.
3. Increment the current value in `left`.

Adding the value to `left` too early could let it pair with itself as a left element. Removing it from `right` too late could let it pair with itself as a right element. The order ensures the three positions are distinct and satisfy `i < j < k`.

### 5. Why multiply the counts?

Suppose there are `2` valid left partners and `3` valid right partners for the current middle value. Each left position can pair with each right position, producing:

\[
2 \times 3 = 6
\]

valid triplets with this middle position. Different indices count as different triplets even when the values are repeated.

### 6. Why a single pass is enough

At every position, `left` summarizes the values before it and `right` summarizes the values after it. This means the algorithm does not need to scan the array again for each possible triplet.

With average constant-time dictionary operations, the approach runs in **expected `O(n)` time** and uses **`O(n)` additional space** in the worst case.

### Summary

Move from left to right, keep counts on both sides of the current position, and add `left[x / r] × right[x × r]` for each valid middle value `x`.

# 8. New Year Chaos

**Topic:** Queues · Inversions · Counting swaps

## Problem overview

People originally stand in the queue in the order `1, 2, 3, ..., n`. A person can bribe the person immediately in front of them to swap positions, while keeping their original number. Each person can bribe **at most two** others.

Given the final queue `q`, print the minimum number of bribes needed to produce it. If any person must have bribed more than two people, print `Too chaotic`.

## HackerRank sample

**Input**

```text
2
5
2 1 5 3 4
5
2 5 1 3 4
```

**Expected output**

```text
3
Too chaotic
```

## Explanation

- For `[2, 1, 5, 3, 4]`, the minimum number of bribes is `3`.
- For `[2, 5, 1, 3, 4]`, person `5` has moved three places forward, which exceeds the allowed limit. The queue is therefore `Too chaotic`.

## Function contract

- **Function shown in this notebook:** `minimumBribes(arr)`
- **Parameter:** `arr` — the final queue order
- **Returns:** no value; the function prints the result or `Too chaotic`

## Key observation

If a person numbered `x` appears more than two positions ahead of their original position, the queue is already invalid under the rules. Otherwise, count the larger-numbered people who appear before each person to determine the number of bribes.

````python
def minimumBribes(arr):
    total = 0
    for i in range(len(arr)):
        if arr[i] - (i + 1) > 2:
            print(f"The person at position {i} has moved more than 2 positions ahead. and the value is {arr[i]}. The result is {arr[i] - (i + 1)}")
            print("Too chaotic")
            return 
        
        for j in range(max(0, arr[i] - 2), i):
            print(f"Checking if arr[j] > arr[i] for j={j} and i={i}: arr[j]={arr[j]}, arr[i]={arr[i]}")
            if arr[j] > arr[i]:
                total += 1
    print(total)
````

````python
arr = [2, 5, 1, 3, 4, 7, 6]
minimumBribes(arr)
````
